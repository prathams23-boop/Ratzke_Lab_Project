# Designing an optimal community: predicting & optimizing pathogen suppression

This self-contained notebook simulates a microbial community with a **genome-based
consumer-resource model (CRM)**, generates data over many community compositions and
nutrient conditions, then uses machine learning to study and **optimize pathogen
suppression** — a first concrete step toward *designing* an optimal community without
understanding its interactions mechanistically.

**It answers three questions:**
1. Can pathogen suppression be predicted from community composition + nutrients?
2. For *large enough* communities, can it be predicted from **nutrients alone**?
3. Can we **optimize the nutrient supply** (using only the ML model) to suppress the pathogen?

Everything runs top-to-bottom with `numpy`, `scipy`, `scikit-learn`, `pandas`, `matplotlib`
(and `joblib`, which ships with scikit-learn). Just *Run All*.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from dataclasses import dataclass, replace
from scipy.integrate import solve_ivp
from scipy.optimize import differential_evolution
from sklearn.ensemble import GradientBoostingRegressor, RandomForestRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score, mean_absolute_error

try:
    from joblib import Parallel, delayed
    HAVE_JOBLIB = True
except Exception:
    HAVE_JOBLIB = False

%matplotlib inline
np.set_printoptions(precision=3, suppress=True)
```

## 1. Configuration

All sizes, rates and sampling ranges live here. Set `QUICK = True` for a fast smoke run
(a few seconds) that exercises the whole pipeline; `QUICK = False` is the full
medium-scale run used for the real analysis (~1–2 min).

```python
QUICK = False  # set True for a fast sanity run

@dataclass(frozen=True)
class Config:
    seed: int = 0
    # community / model dimensions
    n_species: int = 50        # S: species pool (incl. pathogen at index 0)
    n_resources: int = 15      # M: number of nutrients / resources
    genome_length: int = 30    # G: genes per genome
    gene_prob: float = 0.5     # P(gene present) when sampling genomes
    # genome -> parameter mapping
    uptake_scale: float = 1.0
    maint_base: float = 0.05
    maint_cost_per_gene: float = 0.01   # uptake/cost trade-off
    # CRM rates
    growth_conversion: float = 1.0
    dilution_tau: float = 1.0
    # dynamics / integration
    init_biomass: float = 0.1
    t_max: float = 5000.0
    steady_tol: float = 1e-4
    extinction_threshold: float = 1e-6
    # nutrient supply sampling
    nutrient_budget: float = 30.0
    r0_min: float = 0.1
    r0_max: float = 10.0
    # data generation
    n_communities: int = 3000
    richness_min: int = 2
    richness_max: int = 49
    n_jobs: int = -1
    # machine learning
    test_fraction: float = 0.25
    ml_model: str = "gradient_boosting"   # or "random_forest"
    n_richness_bins: int = 6
    large_community_frac: float = 0.7
    # optimisation
    n_baseline_nutrients: int = 200
    de_maxiter: int = 60
    de_popsize: int = 15

cfg = Config()
if QUICK:
    cfg = replace(cfg, n_species=10, n_resources=5, genome_length=12,
                  n_communities=120, richness_max=9, t_max=3000.0,
                  n_richness_bins=3, n_baseline_nutrients=40,
                  de_maxiter=15, de_popsize=8)

PATHOGEN_INDEX = 0   # the pathogen is species 0
print(f"{'QUICK' if QUICK else 'FULL'} run: S={cfg.n_species}, M={cfg.n_resources}, "
      f"G={cfg.genome_length}, communities={cfg.n_communities}")
```

## 2. Genome-based consumer-resource model

Each species carries a binary **genome**. A fixed random *gene → resource* map `B` turns
the genome into a resource-**uptake matrix** `C`. The **maintenance cost** grows with
genome size, so carrying more uptake genes is more expensive — a trade-off that makes
*which* species are present matter. The pathogen is species 0.

Dynamics are a MacArthur CRM with externally supplied (chemostat) resources. For each
present species $i$ and resource $a$:

$$\frac{dN_i}{dt} = g\,N_i\Big(\sum_a C_{ia}\,w_a\,R_a - m_i\Big),\qquad
\frac{dR_a}{dt} = \frac{R^0_a - R_a}{\tau} - \sum_i N_i\,C_{ia}\,R_a$$

- $C_{ia}$ uptake (from the genome), $w_a$ resource quality, $m_i$ maintenance,
  $g$ yield, $R^0_a$ nutrient supply (the optimization knob), $\tau$ turnover time.

We integrate to steady state and record the pathogen's final biomass — **lower = more
suppressed** — as the prediction target.

```python
@dataclass
class Pool:
    genomes: np.ndarray   # (S, G) binary
    B: np.ndarray         # (G, M) gene -> resource map
    C: np.ndarray         # (S, M) uptake matrix
    m: np.ndarray         # (S,)  maintenance cost
    w: np.ndarray         # (M,)  resource quality
    cfg: Config
    @property
    def n_species(self): return self.C.shape[0]
    @property
    def n_resources(self): return self.C.shape[1]

def genome_to_uptake(genomes, B, scale):
    '''Map genomes -> uptake matrix C (S x M). Row-normalised by genome size so
    having more genes spreads capability; the price is paid via maintenance cost.'''
    raw = genomes @ B
    n_genes = np.clip(genomes.sum(axis=1, keepdims=True), 1.0, None)
    return scale * raw / np.sqrt(n_genes)

def build_pool(cfg):
    rng = np.random.default_rng(cfg.seed)
    genomes = (rng.random((cfg.n_species, cfg.genome_length)) < cfg.gene_prob).astype(float)
    for i in range(cfg.n_species):            # ensure no empty genome
        if genomes[i].sum() == 0:
            genomes[i, rng.integers(cfg.genome_length)] = 1.0
    B = rng.gamma(shape=2.0, scale=0.5, size=(cfg.genome_length, cfg.n_resources))
    C = genome_to_uptake(genomes, B, cfg.uptake_scale)
    m = cfg.maint_base + cfg.maint_cost_per_gene * genomes.sum(axis=1)
    w = rng.uniform(0.8, 1.2, size=cfg.n_resources)
    return Pool(genomes, B, C, m, w, cfg)
```

```python
def _rhs(t, y, C_sub, w, m_sub, R0, g, tau):
    k = C_sub.shape[0]
    N, R = y[:k], y[k:]
    dN = g * N * (C_sub @ (w * R) - m_sub)
    consumption = (N[:, None] * C_sub).sum(axis=0)
    dR = (R0 - R) / tau - consumption * R
    return np.concatenate([dN, dR])

def simulate(pool, presence, R0):
    '''Integrate the CRM to steady state for one community (only present species).'''
    cfg = pool.cfg
    presence = np.asarray(presence, dtype=bool)
    idx = np.where(presence)[0]
    N_final = np.zeros(pool.n_species)
    if idx.size == 0:
        return {"N_final": N_final, "R_final": np.asarray(R0, float),
                "reached_steady": True}
    C_sub, m_sub = pool.C[idx], pool.m[idx]
    R0 = np.asarray(R0, float)
    y0 = np.concatenate([np.full(idx.size, cfg.init_biomass), R0.copy()])
    sol = solve_ivp(_rhs, (0.0, cfg.t_max), y0, method="BDF",
                    args=(C_sub, pool.w, m_sub, R0, cfg.growth_conversion, cfg.dilution_tau),
                    rtol=1e-6, atol=1e-9)
    y_end = sol.y[:, -1]
    N_end = np.clip(y_end[:idx.size], 0.0, None)
    R_end = np.clip(y_end[idx.size:], 0.0, None)
    N_end[N_end < cfg.extinction_threshold] = 0.0
    dN = _rhs(sol.t[-1], y_end, C_sub, pool.w, m_sub, R0,
              cfg.growth_conversion, cfg.dilution_tau)[:idx.size]
    alive = N_end > cfg.extinction_threshold
    if alive.any():
        rel = np.abs(dN[alive]) / np.clip(N_end[alive], cfg.extinction_threshold, None)
        reached = bool(rel.max() < cfg.steady_tol)
    else:
        reached = True
    N_final[idx] = N_end
    return {"N_final": N_final, "R_final": R_end,
            "reached_steady": reached and sol.success}

def pathogen_abundance(pool, presence, R0):
    '''Steady-state pathogen biomass (the suppression target). Pathogen forced present.'''
    presence = np.asarray(presence, dtype=bool).copy()
    presence[PATHOGEN_INDEX] = True
    return float(simulate(pool, presence, R0)["N_final"][PATHOGEN_INDEX])
```

### Sanity checks

Quick mechanistic checks that the model behaves: the pathogen grows when alone with
nutrients, gets suppressed in a community, and the state stays non-negative.

```python
pool = build_pool(cfg)
rng = np.random.default_rng(cfg.seed + 99)

def sample_nutrients(rng, cfg):
    '''Nutrient supply R0 drawn log-uniformly per resource, rescaled to the budget.'''
    r0 = np.exp(rng.uniform(np.log(cfg.r0_min), np.log(cfg.r0_max), size=cfg.n_resources))
    return r0 * cfg.nutrient_budget / r0.sum()

alone = np.zeros(cfg.n_species, int); alone[PATHOGEN_INDEX] = 1
r0 = sample_nutrients(rng, cfg)
n_alone = pathogen_abundance(pool, alone, r0)

others = [i for i in range(cfg.n_species) if i != PATHOGEN_INDEX]
comm = alone.copy()
comm[rng.choice(others, size=min(20, len(others)), replace=False)] = 1
n_comm = pathogen_abundance(pool, comm, r0)

print(f"pathogen alone:            {n_alone:8.3f}  ({'OK' if n_alone>0 else 'FAIL'})")
print(f"pathogen in 20-sp comm:    {n_comm:8.3f}  (suppression "
      f"{'observed' if n_comm < n_alone else 'NOT observed'})")
```

## 3. Generate the dataset

Each sample: pick a community *richness* (number of non-pathogen species, spanning
small → large), choose that many species at random (pathogen always present), draw a
nutrient supply vector under a fixed total budget, simulate to steady state, and record
the pathogen's final biomass.

```python
SP_COLS = [f"sp_{i}" for i in range(cfg.n_species)]
R0_COLS = [f"r0_{a}" for a in range(cfg.n_resources)]

def sample_composition(rng, cfg):
    presence = np.zeros(cfg.n_species, int)
    presence[PATHOGEN_INDEX] = 1
    others = [i for i in range(cfg.n_species) if i != PATHOGEN_INDEX]
    richness = min(int(rng.integers(cfg.richness_min, cfg.richness_max + 1)), len(others))
    presence[rng.choice(others, size=richness, replace=False)] = 1
    return presence

def generate_dataset(pool, cfg):
    rng = np.random.default_rng(cfg.seed + 1)
    comps = np.empty((cfg.n_communities, cfg.n_species), int)
    nutr = np.empty((cfg.n_communities, cfg.n_resources), float)
    for k in range(cfg.n_communities):
        comps[k] = sample_composition(rng, cfg)
        nutr[k] = sample_nutrients(rng, cfg)

    def run_one(k):
        out = simulate(pool, comps[k], nutr[k])
        return out["N_final"][PATHOGEN_INDEX], out["reached_steady"]

    if HAVE_JOBLIB:
        try:
            res = Parallel(n_jobs=cfg.n_jobs)(delayed(run_one)(k)
                                              for k in range(cfg.n_communities))
        except Exception:
            res = [run_one(k) for k in range(cfg.n_communities)]
    else:
        res = [run_one(k) for k in range(cfg.n_communities)]
    pathogen, reached = zip(*res)

    df = pd.DataFrame(comps, columns=SP_COLS)
    df[R0_COLS] = nutr
    df["richness"] = comps.sum(axis=1) - 1          # exclude the pathogen
    df["reached_steady"] = np.asarray(reached, bool)
    df["pathogen_abundance"] = np.asarray(pathogen, float)
    return df

df = generate_dataset(pool, cfg)
print(f"{len(df)} communities | {df.reached_steady.mean():.0%} reached steady state")
print(f"pathogen abundance: mean={df.pathogen_abundance.mean():.2f}  "
      f"range=[{df.pathogen_abundance.min():.2f}, {df.pathogen_abundance.max():.2f}]")
df[["richness"] + R0_COLS[:3] + ["pathogen_abundance"]].head()
```

## 4. Q1 — Can we predict pathogen suppression?

Train a **full** model (composition + nutrients) and a **nutrients-only** model, and
report holdout $R^2$.

```python
def make_model(cfg):
    if cfg.ml_model == "random_forest":
        return RandomForestRegressor(n_estimators=300, n_jobs=cfg.n_jobs,
                                     random_state=cfg.seed)
    return GradientBoostingRegressor(random_state=cfg.seed)

def fit_eval(df, feat_cols, cfg):
    X = df[feat_cols].to_numpy(); y = df["pathogen_abundance"].to_numpy()
    Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=cfg.test_fraction,
                                          random_state=cfg.seed)
    model = make_model(cfg).fit(Xtr, ytr)
    pred = model.predict(Xte)
    return model, yte, pred, r2_score(yte, pred), mean_absolute_error(yte, pred)

full_model, yte, pred, full_r2, full_mae = fit_eval(df, SP_COLS + R0_COLS, cfg)
_, _, _, nutr_r2, nutr_mae = fit_eval(df, R0_COLS, cfg)
print(f"full model (composition + nutrients): R2={full_r2:.3f}  MAE={full_mae:.2f}")
print(f"nutrients-only model (all data):      R2={nutr_r2:.3f}  MAE={nutr_mae:.2f}")

fig, ax = plt.subplots(figsize=(5.2, 5.2))
ax.scatter(yte, pred, s=12, alpha=0.4)
lim = [min(yte.min(), pred.min()), max(yte.max(), pred.max())]
ax.plot(lim, lim, "k--", lw=1)
ax.set_xlabel("true pathogen abundance"); ax.set_ylabel("predicted")
ax.set_title(f"Full model parity ($R^2$={full_r2:.3f})")
plt.show()
```

## 5. Q2 — For large communities, do nutrients alone suffice?

This is a *within-regime* question: among communities of a given size, does adding
composition improve prediction? So we train **both models inside each richness bin** —
both on the same data, so the **gap between them is a fair test**. The gap closing at
high richness means composition stops carrying extra information once the community is
diverse.

We also plot a **reference "ceiling"**: the full model trained on *all* data and scored
per bin. (A pooled *nutrients-only* model would be unfair — it never sees the richness
regime — so only the full model is used as a reference.) The per-bin curves sit below
this ceiling mainly because each is trained on a fraction of the data, **not** because
large communities are intrinsically harder to predict.

```python
def richness_analysis(df, cfg):
    full_cols = SP_COLS + R0_COLS
    y = df["pathogen_abundance"].to_numpy(); rich = df["richness"].to_numpy()
    idx = np.arange(len(df))
    i_tr, i_te = train_test_split(idx, test_size=cfg.test_fraction, random_state=cfg.seed)
    train_mask = np.zeros(len(df), bool); train_mask[i_tr] = True
    Xf = df[full_cols].to_numpy(); Xn = df[R0_COLS].to_numpy()

    ref_full = make_model(cfg).fit(Xf[i_tr], y[i_tr])   # reference ceiling

    edges = np.linspace(rich.min(), rich.max() + 1, cfg.n_richness_bins + 1)
    rows = []
    for lo, hi in zip(edges[:-1], edges[1:]):
        in_bin = (rich >= lo) & (rich < hi)
        btr = np.where(in_bin & train_mask)[0]
        bte = np.where(in_bin & ~train_mask)[0]
        if len(btr) < 16 or len(bte) < 8 or np.ptp(y[bte]) == 0:
            continue
        mf = make_model(cfg).fit(Xf[btr], y[btr])
        mn = make_model(cfg).fit(Xn[btr], y[btr])
        rows.append(dict(richness_mid=0.5*(lo+hi), n=len(bte),
                         r2_full=r2_score(y[bte], mf.predict(Xf[bte])),
                         r2_nutrients=r2_score(y[bte], mn.predict(Xn[bte])),
                         r2_full_ref=r2_score(y[bte], ref_full.predict(Xf[bte]))))
    return pd.DataFrame(rows)

rich_df = richness_analysis(df, cfg)
display(rich_df.round(3))

fig, ax = plt.subplots(figsize=(7.5, 5))
ax.plot(rich_df.richness_mid, rich_df.r2_full, "o-",
        label="full: composition + nutrients (per bin)")
ax.plot(rich_df.richness_mid, rich_df.r2_nutrients, "s--",
        label="nutrients only (per bin)")
ax.plot(rich_df.richness_mid, rich_df.r2_full_ref, "^:", color="gray", alpha=0.8,
        label="full model trained on all data (reference ceiling)")
ax.set_xlabel("community richness (# non-pathogen species)")
ax.set_ylabel(r"holdout $R^2$ for pathogen abundance")
ax.set_title("Predictability vs community size\n"
             "(per-bin gap closing => nutrients alone suffice when diverse)")
ax.legend(); ax.grid(alpha=0.3); plt.show()

g0 = rich_df.iloc[0].r2_full - rich_df.iloc[0].r2_nutrients
g1 = rich_df.iloc[-1].r2_full - rich_df.iloc[-1].r2_nutrients
print(f"full - nutrients gap:  {g0:.3f} (small communities) -> {g1:.3f} (large communities)")
print("=> gap shrinking toward 0 means composition adds nothing extra once diverse.")
```

### Which features matter?

Feature importances of the full model (orange = nutrients, blue = species presence).

```python
if hasattr(full_model, "feature_importances_"):
    names = np.array(SP_COLS + R0_COLS)
    imp = full_model.feature_importances_
    order = np.argsort(imp)[::-1][:20]
    colors = ["tab:orange" if n.startswith("r0_") else "tab:blue" for n in names[order]]
    fig, ax = plt.subplots(figsize=(7, 0.35*len(order)+1))
    ax.barh(range(len(order)), imp[order], color=colors)
    ax.set_yticks(range(len(order))); ax.set_yticklabels(names[order]); ax.invert_yaxis()
    ax.set_xlabel("feature importance")
    ax.set_title("Full model: top features (orange = nutrient, blue = species)")
    plt.show()
```

## 6. Q3 — Optimize nutrients to suppress the pathogen

Train a **nutrients-only surrogate** on the *large-community* subset, optimize the
nutrient supply (under the same total budget) to minimize the surrogate's predicted
pathogen abundance with `differential_evolution`, then **validate on the true simulator**:
fix a representative large community where the pathogen still survives and compare the
optimized nutrients against many random nutrient draws.

```python
def project_to_budget(r0, cfg):
    r0 = np.clip(r0, cfg.r0_min, cfg.r0_max)
    return r0 * cfg.nutrient_budget / r0.sum()

# 1) surrogate on the large-community subset
threshold = cfg.large_community_frac * cfg.n_species
large = df[df.richness >= threshold]
if len(large) < 16: large = df
surrogate = make_model(cfg).fit(large[R0_COLS].to_numpy(),
                                large["pathogen_abundance"].to_numpy())

# 2) optimize nutrients on the surrogate (abundance can't be negative -> clip)
def objective(r0):
    r0p = project_to_budget(np.asarray(r0), cfg)
    return max(0.0, float(surrogate.predict(r0p.reshape(1, -1))[0]))
res = differential_evolution(objective, [(cfg.r0_min, cfg.r0_max)] * cfg.n_resources,
                             maxiter=cfg.de_maxiter, popsize=cfg.de_popsize,
                             seed=cfg.seed, polish=True, tol=1e-7)
r0_opt = project_to_budget(res.x, cfg)
surrogate_pred = res.fun

# 3) pick a large community where the pathogen survives (room to optimize)
sub = df[df.richness >= threshold]
if len(sub) == 0: sub = df[df.richness >= df.richness.quantile(0.9)]
presence = df.loc[sub.pathogen_abundance.idxmax(), SP_COLS].to_numpy(int)
presence[PATHOGEN_INDEX] = 1

# 4) validate on the TRUE simulator vs random nutrient baselines
rng = np.random.default_rng(cfg.seed + 7)
baseline = np.array([pathogen_abundance(pool, presence, sample_nutrients(rng, cfg))
                     for _ in range(cfg.n_baseline_nutrients)])
optimised = pathogen_abundance(pool, presence, r0_opt)

print(f"optimized on a richness-{int(presence.sum()-1)} community")
print(f"surrogate prediction:           {surrogate_pred:.3f}")
print(f"true CRM at optimized nutrients: {optimised:.3f}")
print(f"baseline (random) mean:          {baseline.mean():.3f}  (best random {baseline.min():.3f})")

fig, ax = plt.subplots(figsize=(7.5, 5))
ax.hist(baseline, bins=30, alpha=0.7, color="tab:blue",
        label="random nutrient conditions (true CRM)")
ax.axvline(baseline.mean(), color="tab:blue", ls=":", label=f"baseline mean = {baseline.mean():.2f}")
ax.axvline(optimised, color="tab:red", lw=2, label=f"optimized (true CRM) = {optimised:.2f}")
ax.axvline(surrogate_pred, color="tab:green", ls="--", lw=2,
           label=f"surrogate prediction = {surrogate_pred:.2f}")
ax.set_xlabel("pathogen abundance (lower = better suppression)")
ax.set_ylabel("count"); ax.set_title("Nutrient optimization validated on the true simulator")
ax.legend(fontsize=8); plt.show()
```

## 7. Takeaways & next steps

- **Q1:** pathogen suppression is predictable from composition + nutrients (full-model $R^2$).
- **Q2:** the per-bin full-vs-nutrients gap shrinks toward 0 as richness grows — for large,
  diverse communities **nutrients alone predict suppression** about as well as knowing the
  full composition.
- **Q3:** optimizing the nutrient supply on the ML surrogate and re-running the true CRM
  drives the pathogen well below the random-nutrient baseline.

**Natural extensions** (per the original project idea): an *ensemble* surrogate that yields
uncertainty $\sigma$, an acquisition function $\mu + b\,\sigma$ (exploration vs exploitation),
and an active-learning loop that simulates new communities chosen by the optimizer —
i.e. Bayesian optimization / metaheuristics over the binary composition space.
