```python
# Recursion for printing fibonacci sequence
# always remember to mention base case while creating recursion functions
# formula for fibonacci: fib(n)= fib(n-2)+fib(n-1)

def fib(n):
    # base case:
    if(n==0 or n==1):
        return n
        
    return fib(n-2)+fib(n-1)

fib(0)
```

```python
# recursive function for factorial

def fact(n):
    if n==0 or n==1:
        return n
    return n*fact(n-1)

print(fact(0))
print(fact(1))
print(fact(5))
```

```python
# function for sum of all digits of a given number

def sum_of_digits(n):

    x=str(n)
    ans=0
    for i in range(0, len(x)):
        ans+=int(x[i])
    return ans

print(sum_of_digits(154445626596261464431236461313))
print(sum_of_digits(6))
print(sum_of_digits(44))
print(sum_of_digits(0))
            
    
```

```python
# find product with highest prices given relevant dictionary

products = {
    "apple": 120,
    "banana": 40,
    "mango": 150,
    "orange": 90
}

highest_product = max(products, key=products.get)

print("Product with highest price:", highest_product)
print("Price:", products[highest_product])

```
