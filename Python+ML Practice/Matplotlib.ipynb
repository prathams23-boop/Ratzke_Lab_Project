```python
from matplotlib import pylab
import numpy as np
```

```python
# Using numpy to generate random data

x=np.linspace(0,10,25)
y=(x*x)+2
print(x)
print(y)

pylab.plot(x,y,'r') # r stands for red


```

```python
import matplotlib.pyplot as plt
```

```python
x=np.arange(0,8)
y=np.sin(x)
print(x)
print(y)

plt.plot(x,y)
plt.xlabel('x')
plt.ylabel('sinx')
plt.show()
```

```python
# A better sine plot

xstart=0
xstop=2*np.pi
increment=0.1

x=np.arange(xstart,xstop,increment)
y=np.sin(x)

plt.plot(x,y)
plt.xlabel('x')
plt.ylabel('sinx')
plt.show()
```

# Subplots

```python
''' The subplot command enables to display multiple plots in the same window
    subplot(m,n,p) partitions the figure window into:
    m by n matrix with p being the indexes'''

xstart=0
xstop=2*np.pi
increment=0.1

x=np.arange(xstart,xstop,increment)
y=np.sin(x)
z=np.cos(x)

plt.subplot(2,1,1)
plt.plot(x,y,'g')
plt.title('Sine curve')
plt.xlabel('x')
plt.ylabel('sin(x)')
plt.grid()
plt.legend(['y=sin(x)'], loc=1)
plt.show


plt.subplot(2,1,2)
plt.plot(x,z,'r')
plt.title('Cosine curve')
plt.xlabel('x')
plt.ylabel('cosx)')
plt.grid()
plt.legend(['y=cos(x)'], loc=1)
plt.show




```
