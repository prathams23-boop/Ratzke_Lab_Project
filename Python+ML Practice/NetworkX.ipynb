```python
# Import libraries 
import networkx as nx # Network analysis
import matplotlib.pyplot as plt # Data visualization
import pydot # Python interface to Graphviz
from networkx.drawing.nx_pydot import graphviz_layout
# Create a graph
graph = nx.Graph() 
```

```python
# Suppose we have 4 individuals: A,B,C,D
# Add the nodes

graph.add_nodes_from(['A','B','C','D'])

# The friendship connections are as follows
'''A is friends with B and C
   B is friends with A and D
   C is friends with A
   D is friends with B'''

# Adding the edges

graph.add_edges_from([('A','B'),('A','C'),('B','D'),('B','A')])

# Now we can draw and visualise the graph

pos = graphviz_layout(graph, prog="dot")

nx.draw(graph, 
        pos, 
        with_labels=True, 
        node_size=1000, 
        node_color=["pink", "yellow", "tan", "orange"])
plt.show()

```
