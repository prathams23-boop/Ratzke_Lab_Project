# Series creation, manipulation, querying and deletion

```python
arr=[0,1,2,3,4]
s1=pd.Series(arr) # list has been converted to series
print(s1)

print('Changing the Index')
order=[1,2,3,4,5]
s2=pd.Series(arr, index=order)
print(s2)
# Or we can directly change the index
s2.index=['a','b','c','d','e']
print(s2)

# Creating a series from a dictionary

print('Series from Dictionary')
d={'a':1, 'b':2, 'c':3, 'd':4}
s3=pd.Series(d)
print(s3)
```

```python
s4=pd.concat([s1,s3])
print(s4)
print(s4.drop('d'))
```

# Series Operations



```python
arr1=[1,2,3,4,5,6,7]
arr2=[1,2,3,4]
ser1=pd.Series(arr1)
ser2=pd.Series(arr2)
print(ser1.add(ser2))
print(ser1.sub(ser2))

# Similar for multiply (mul) and divide (div)
```

```python
print(ser1)
print('Median is:',ser1.median())
print('Minimum value is:',ser1.min())
print('Maximum value is:',ser1.max())
```

# Operations for missing values in DataFrames

```python
# Create dictionary with NaN values
data = {
    "Age": [22, 25, np.nan, 28, 35],
    "Salary": [30000, np.nan, 50000, 45000, 60000],
    "Score": [75, 80, 85, np.nan, 90]
}

# Create DataFrame from dictionary
dfnan = pd.DataFrame(data)

# Print original DataFrame
print("Original DataFrame:\n", dfnan)

# Describe statistics (NaNs are automatically ignored)
print("\nDescribe:\n", dfnan.describe())

# Calculate mean of each column (NaNs ignored automatically)
meanAge = dfnan['Age'].mean()
meanSal = dfnan['Salary'].mean()
meanScore = dfnan['Score'].mean()

# Fill NaN values and ASSIGN BACK 
dfnan['Age'] = dfnan['Age'].fillna(meanAge)
dfnan['Salary'] = dfnan['Salary'].fillna(meanSal)
dfnan['Score'] = dfnan['Score'].fillna(meanScore)

# Print updated DataFrame
print("\nDataFrame after filling NaNs:\n", dfnan)
```
