```python
# =====================================================
# Example 1: Simple 3D array
# =====================================================

# Create a 3D array with numbers from 0 to 23
# reshape(2,3,4) means:
# axis 0 → 2 blocks
# axis 1 → 3 rows
# axis 2 → 4 columns

a = np.arange(24).reshape(2,3,4)

print("Original array:")
print(a)

print("\nOriginal shape:", a.shape)
# (2,3,4)

# Move axis 2 to the front (position 0)
b = np.rollaxis(a, 2)

print("\nAfter np.rollaxis(a,2)")
print("New shape:", b.shape)

# Explanation:
# original axes: 0,1,2
# move axis 2 → front
# new order: 2,0,1
# shape: (4,2,3)



# Move axis 2 to position 1
c = np.rollaxis(a, 2, 1)

print("\nAfter np.rollaxis(a,2,1)")
print("New shape:", c.shape)

# Explanation:
# original axes: 0,1,2
# move axis 2 → position 1
# new order: 0,2,1
# shape: (2,4,3)



# =====================================================
# Example 2: Larger array with more axes
# =====================================================

# Create a 4D array
# reshape(2,3,4,5)
# axis 0 → size 2
# axis 1 → size 3
# axis 2 → size 4
# axis 3 → size 5

big_array = np.arange(120).reshape(2,3,4,5)

print("\n\nOriginal 4D array shape:", big_array.shape)

# -----------------------------------------------------
# Move axis 3 to the front
# -----------------------------------------------------

rolled1 = np.rollaxis(big_array, 3)

print("\nAfter rollaxis(big_array,3)")
print("New shape:", rolled1.shape)

# Explanation:
# original axes: 0,1,2,3
# move axis 3 → front
# new axes: 3,0,1,2
# new shape: (5,2,3,4)



# -----------------------------------------------------
# Move axis 3 to position 1
# -----------------------------------------------------

rolled2 = np.rollaxis(big_array, 3, 1)

print("\nAfter rollaxis(big_array,3,1)")
print("New shape:", rolled2.shape)

# Explanation:
# original axes: 0,1,2,3
# move axis 3 → position 1
# new axes: 0,3,1,2
# new shape: (2,5,3,4)



# -----------------------------------------------------
# Move axis 2 to position 1
# -----------------------------------------------------

rolled3 = np.rollaxis(big_array, 2, 1)

print("\nAfter rollaxis(big_array,2,1)")
print("New shape:", rolled3.shape)

# Explanation:
# original axes: 0,1,2,3
# move axis 2 → position 1
# new axes: 0,2,1,3
# new shape: (2,4,3,5)
```

# NaN

```python
# ============================================
# Working with Missing Values (NaN) in NumPy
# ============================================

# ------------------------------------------------
# Step 1: Create a NumPy array containing NaN values
# ------------------------------------------------

# np.nan represents "Not a Number" (missing value)
arr = np.array([10, 20, np.nan, 40, np.nan, 60])

print("Original Array:")
print(arr)


# ------------------------------------------------
# Step 2: Detect which elements are NaN
# ------------------------------------------------

# np.isnan() checks each element and returns a boolean array
# True  -> element is NaN
# False -> element is not NaN

nan_mask = np.isnan(arr)

print("\nBoolean mask showing NaN locations:")
print(nan_mask)


# ------------------------------------------------
# Step 3: Count the total number of NaN values
# ------------------------------------------------

# True is treated as 1 and False as 0
# So summing the boolean array gives the count of NaN values

nan_count = np.isnan(arr).sum()

print("\nTotal number of NaN values:")
print(nan_count)


# ------------------------------------------------
# Step 4: Find the indices (positions) of NaN values
# ------------------------------------------------

# np.where() returns the indices where the condition is True
nan_indices = np.where(np.isnan(arr))

print("\nIndices of NaN values:")
print(nan_indices)


# ------------------------------------------------
# Step 5: Access the elements at those indices
# ------------------------------------------------

# Using the boolean mask to extract NaN values
missing_values = arr[np.isnan(arr)]

print("\nExtracted NaN values from the array:")
print(missing_values)


# ------------------------------------------------
# Step 6: Example with a 2D array
# ------------------------------------------------

arr2 = np.array([
    [1, 2, np.nan],
    [4, np.nan, 6],
    [7, 8, 9]
])

print("\n\n2D Array:")
print(arr2)


# Detect NaN positions
nan_mask_2d = np.isnan(arr2)

print("\nNaN mask for 2D array:")
print(nan_mask_2d)


# Get row and column indices of NaN values
nan_indices_2d = np.where(np.isnan(arr2))

print("\nIndices of NaN values in 2D array (row indices, column indices):")
print(nan_indices_2d)
```
